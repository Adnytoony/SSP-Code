```python
 #calculates for the amount of days, hours, and minutes since June 1, 2026 at noon UTC
def julian_date(day, hours, minutes):
    start_JD = 2461192.5 #June 1, 2026 at noon Julian Date (from presentation)
    #Rises by 1 each day
    while day > 1: #for every full day since June 1, add 1 to the julian date
            start_JD +=1
            day -=1 #subtract a day from the given
    while hours >= 1: #for every full hour since June 1, add 1/24 to the julian date
            start_JD += (1/24)
            hours -=1 #subtract an hour from the given
    while minutes >= 1: #for every minutes since June 1, add (1/24)/60 to the julian date
            start_JD += (1/(24*60))
            minutes -=1 #subtract a minute from the given

    return start_JD #returns all the added numbers

print("The Julian Date for June 17, 2026 at 2:59 UTC is ", julian_date(19, 2, 59))
print("Expected: 2461210.624305556")
            
```
