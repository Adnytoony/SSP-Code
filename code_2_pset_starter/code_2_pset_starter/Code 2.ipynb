# Problem 1: Plotting a trajectory

```python
# -- Plot the trajectory of the ball --
import numpy as np
import matplotlib.pyplot as plt

# YOUR CODE HERE
t_values = np.linspace(0, 5.4, 100)

y_values = (-3.71*(t_values)**2)/2 + 10*(t_values)
plt.figure(figsize=(10, 6))

# Plot our two curves (the x_values and y_values we computed earlier)
plt.plot(t_values, y_values, label='y(t) = (-3.71t^2)/2 + 10t')

# Add a legend that labels what each curve is
plt.legend()

# Add a title and axis labels
plt.title('Astronaut throws ball in Mars')
plt.xlabel('Time')
plt.ylabel('Height')

# Display the plot! Prevents further changes
plt.show()
```
